cuando el contenido de la columna no es identificado se coloca de tipo object por o tambien cuando la info es de tipo texto esto lo realiza de manera automatica Python acontinuacion se van a trabajar para asignarle el tipo correcto a cada dato

```python
import numpy as np
```

```python
datos['max_hospedes']= datos['max_hospedes'].astype(np.int64)
```

en esta parte se selecciona la columna datos['max_hospedes'] y se le aplica lla funcion de la libreria numpy para asignar valores en este caso se especifica que se va a trabajar con datos de tipo entero si en algun valor de la columna no es entero esto no va a funcionar

```python
columnas_numericas = ['cantidad_baños','cantidad_cuartos', 'cantidad_camas']
```

```python
datos[columnas_numericas]= datos[columnas_numericas].astype(np.int64)
```

```python
datos.info()
```

```python
datos['evaluacion_general']= datos['evaluacion_general'].astype(np.float64)
```

```python
datos['precio']=datos['precio'].apply(lambda x: x.replace('$','').replace(',','').strip())
```

en este caso como la columna precio tenia un caracter de signo pesos es necesario hacerle una tranformacion para pposterior asignar un tipo de dato corecto para esto se utilizo la funcion lamda que lo que hace es recorrer toda la columna y aplicrale un funcion que definimos mas adelante en este caso la llamamos "X" que es el remplazar ese simbolo con nada y en el  ejemplono solo hicimos una tranformacion se realizaron tres tranformacionos la primera quita el signo pesos de los registros que encuentre en la segunda decimos si encuentra una , retirela (python reconoce y trabaja bien con los valores flotantes separados por . no por , por eso esa transformacion y por ultimo se aplica .strip que lo que hace es eliminar los espacios ya sea al inicio o al final)

```python
datos['precio']= datos['precio'].astype(np.float64)
```

```python
datos[['cuota_deposito','cuota_limpieza']] = datos[['cuota_deposito','cuota_limpieza']].map(lambda x: x.replace('$','').replace(',','').strip())
```

cuando es una unica columna tenemos una serie si tenemos mas de una columna ya python lo trata como un dataframe por eso es necesario cambiar como se se aplica las tranformaciones que estan diseñadas solo para aplicar sobre series por eso el applymap

```python
datos[['cuota_deposito','cuota_limpieza']]=datos[['cuota_deposito','cuota_limpieza']].astype(np.float64)
```

# datos de texto
vamos a tokenizar cada palabra va a ser un token

```python
datos['descripcion_local']=datos['descripcion_local'].str.lower() #se tramforma los valores en tipo texto y se les aplica la funcion lower que devuelve el texto en minuscula
```

```python
datos['descripcion_local']=datos['descripcion_local'].str.replace('[^a-zA-Z0-9\-\']',' ',regex=True)#en esta parte le estoy pidiendo que me replace todo menso [^a-zA-Z0-9\-\'] esa negacion se logra con el ^
```

```python
datos['descripcion_local']= datos['descripcion_local'].str.replace('(?<!\w)-(?!\w)',' ',regex=True)
```

```python
datos['descripcion_local'].head(10)
```

```python
datos['descripcion_local']= datos['descripcion_local'].str.split()#en este caso el split separa cada palabra y devuelve una lista por defecto el separador es el espacio
```

```python
datos['comodidades']= datos['comodidades'].str.replace('\{|}]|\"','',regex=True)
#en este caso se pide con el \ que busque algo en especifico y el | me dice es ademas busca este otro signo
```

```python
datos['comodidades']= datos['comodidades'].str.split(',')
datos['comodidades'].head()
```
